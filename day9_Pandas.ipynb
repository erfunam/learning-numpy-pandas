```python
# create a new dataframe with transaction_amount, ip_risk_score, and payment_method columns
gdf = pd.DataFrame(
    {
        'transaction_amount' : [50, 1500, 200, 3000],
        'ip_risk_score' : [0.1, 0.9, 0.3, 0.8],
        'payment_method' : ['credit_card', 'paypal', 'credit_card', 'paypal']
    }
)
gdf
```

```python
# statistical summary of the dataframe for numeric columns
gdf.describe(include= 'all')
```

```python
# group by payment_method and calculate the mean of transaction_amount for each group
mean_result = gdf.groupby('payment_method')['transaction_amount'].mean()
print(mean_result)
```

```python
# group by payment_method and calculate the mean and count of transaction_amount for each group
mc_result = gdf.groupby('payment_method')['transaction_amount'].agg(['mean', 'count'])
print(mc_result)
```

```python
# group by payment_method and calculate the mean of ip_risk_score for each group
iprisk_gdf = gdf.groupby('payment_method')['ip_risk_score'].mean()
print(iprisk_gdf)
```

```python
# group by payment_method and calculate the mean and count of ip_risk_score for each group
mcIprisk_gdf = gdf.groupby('payment_method')['ip_risk_score'].agg(['mean', 'count'])
print(mcIprisk_gdf)
```

```python
# group by payment_method and calculate the mean of all numeric columns for each group
gdf.groupby('payment_method').mean()
```

### create new columns
### basic feature engineering

```python
new_df = pd.DataFrame(
    {
        'transaction_amount' : [50, 1500, 200, 3000],
        'ip_risk_score' : [0.1, 0.9, 0.2, 0.8]
    }
)

new_df
```

```python
new_df['amount_log'] = np.log(new_df['transaction_amount']) # add a new column with the log of transaction_amount
new_df
```

```python
new_df['is_expensive'] = new_df['transaction_amount'] > 1000 # add a new column with boolean values indicating if the transaction is expensive
new_df
```

```python
new_df['risk_category'] = new_df['ip_risk_score'].apply(
    lambda x : 'high' if x > 0.6 else 'low'
)
new_df
```

### merg

```python
merged_df = pd.merge(gdf, new_df, on= 'transaction_amount') # merge two dataframes on transaction_amount column
merged_df
```

```python
new_merged_df = gdf.merge(new_df, on= 'transaction_amount', how= 'left') # merge two dataframes on transaction_amount column with left join
new_merged_df
```

### working with dates

```python
gdf
```

```python
date = ['2024-03-15 08:30:00', '2024-03-16 23:45:00', '2024-03-22 10:15:00', '2024-03-23 14:10:00']
gdf['transaction_date'] = date # add a new column with transaction dates
print(gdf)
print(gdf.dtypes) # print the data types of each column
```

```python
# convert the transaction_date column to datetime format
gdf['transaction_date'] = pd.to_datetime(gdf['transaction_date'])

print(gdf)
print(gdf.dtypes) # print the data types of each column
```

```python
gdf['hour'] = gdf['transaction_date'].dt.hour # add a new column with the hour of the transaction
gdf['day_of_week'] = gdf['transaction_date'].dt.dayofweek # add a new column with the day of the week of the transaction (0=Monday, 6=Sunday)
gdf['day_name'] = gdf['transaction_date'].dt.day_name() # add a new column with the name of the day of the week of the transaction
gdf['month'] = gdf['transaction_date'].dt.month # add a new column with the month of the transaction
gdf['date_only'] = gdf['transaction_date'].dt.date # add a new column with the date of the transaction

gdf
```

```python
fdf = pd.DataFrame({
    'pickup': ['2024-03-15 08:00:00', '2024-03-16 10:00:00'],
    'dropoff': ['2024-03-15 08:25:00', '2024-03-17 10:50:00']
})
fdf['pickup'] = pd.to_datetime(fdf['pickup'])
fdf['dropoff'] = pd.to_datetime(fdf['dropoff'])

fdf['duration'] = fdf['dropoff'] - fdf['pickup'] # add a new column with the duration of the trip
fdf
```

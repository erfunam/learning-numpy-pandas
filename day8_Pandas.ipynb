### some function and method

```python
print(df.head()) # first 5 rows
print(df.head(10)) # first 10 rows
print(df.tail()) # last 5 rows
print(df.info()) # information about the DataFrame
print(df.describe()) # statistical summary of the DataFrame
print(df.shape) # shape of the DataFrame (rows, columns)
print(df.columns) # column names
```

### selecting columns and rows

```python
print(df['A']) # select column A
print(df[['A']]) # select column A as a DataFrame
print(df[['A', 'B']]) # select columns A and B as a DataFrame

print(df[0:3]) # select first 3 rows
print(df.loc['2023-01-01']) # select row by label
print(df.loc[dates[0]]) # select row by label
print(df.loc[:, ['A', 'B']]) # select all rows and columns A and B
print(df.loc['2023-01-01', ['A', 'B']]) # select row by label and columns A and B
print(df.iloc[3]) # select row by integer position
print(df.iloc[3:5, 0:2]) # select rows 3 to
```

# import pandas

```python
import pandas as pd
```

### create data frame

```python
df = pd.DataFrame (
    {
        "Name": ['erfan', 'parsa', 'sirous'],
        "Age": [20, 22, 40],
        "Gender": ['male', 'male', 'male'],
        "Ghad": [180, 170, 195],
    }
)

print(df)
```

### create series (one columns)

```python
mive = pd.Series(['orange', 'apple', 'banana'], name = 'mive')

print(mive)
```

### some functions

```python
print(df["Age"].max())
print(df["Age"].min())
print(df["Age"].mean())
print(df["Age"].std())
```

### general discribe

```python
df.describe()
```

```python
df.describe(include= "all")
```

```python
df.describe(include= "object")
```

```python
df.describe(include= "number")
```

# import numpy

```python
import numpy as np
```

### generating random number with numpy

```python
s = pd.Series(np.random.randn(1000))

s.describe()
```

### filtering data with nan

```python
s[::2] = np.nan

s.describe()
```

### create data frame with numpy random number

```python
ss = pd.DataFrame(np.random.randn(1000, 5), columns= ['a', 'b', 'c', 'd', 'e'])

ss
```

```python
ss.describe()
```
